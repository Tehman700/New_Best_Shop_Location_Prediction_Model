```python
import pandas as pd

# URL of the page with population table
url = 'https://www.citypopulation.de/en/pakistan/admin/islamabad/501__islamabad/'

# Use pandas to read all tables from the webpage
tables = pd.read_html(url)

print(f"Found {len(tables)} tables.")

# Inspect first table
pop_df = tables[0]

# Show columns and head to confirm
print(pop_df.columns)
print(pop_df.head())

# Clean columns or filter as needed
# Then save to CSV
pop_df.to_csv('population_data.csv', index=False)

print("✅ Population data saved as population_data.csv")
```
